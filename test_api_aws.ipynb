# Test the API on AWS

Same call as `test_api_local.ipynb`: POST the first selected row from `artifacts/payload_test.json` to `/predict`.

Use the `PredictUrl` value from the CloudFormation stack Outputs tab. The stack must be `CREATE_COMPLETE`, and port 8000 must be open to your IP.

```python
import json
from pathlib import Path


path_data = Path('artifacts/payload_test.json')

with open(
    path_data,
    "r",
    encoding="utf-8"
) as f:
    data = [json.loads(line) for line in f if line.strip()]


X = data[4]
print('Using X =', X)
```

```python
import requests

# Paste PredictUrl from the stack Outputs tab.
url = 'http://13.63.45.126:8000/predict'

try:
    r = requests.post(url, json={'X': X}, timeout=30)
    print('Status:', r.status_code)
    print('Response:', r.json())
except Exception as e:
    print('Request failed:', e)
```
