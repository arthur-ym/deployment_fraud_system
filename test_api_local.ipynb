# Test API Notebook
Calls the local FastAPI `/predict` endpoint using the first element from `artifacts/payload_test.json`.

```python
import json
from pathlib import Path


path_data = Path('artifacts/payload_test.json')

with open(
    path_data,
    "r",
    encoding="utf-8"
) as f:
    data = [json.loads(line) for line in f if line.strip()]


X=data[4]
print('Using X =', X)
```

```python
import requests

url = 'http://localhost:8000/predict'
try:
    r = requests.post(url, json={'X': X}, timeout=10)
    print('Status:', r.status_code)
    print('Response:', r.json())
except Exception as e:
    print('Request failed:', e)
```
