Calls the local FastAPI `/predict` endpoint using the first element from `artifacts/payload_test.json`.

```python
from pathlib import Path
import json

path_data = Path('artifacts/payload_test.json') if Path('artifacts/payload_test.json').exists() else Path('artifacts/payload_test.json')

with open(
    path_data,
    "r",
    encoding="utf-8"
) as f:
    data = [json.loads(line) for line in f if line.strip()]

if isinstance(data, list):
    first = data[0]
elif isinstance(data, dict):
    # prefer key 'X' if present
    if 'X' in data:
        first = data['X']
    else:
        # take first value from the dict
        first = next(iter(data.values()))
else:
    first = data

X=values = list(first.values())
print('Using X =', X)
```

```python
import requests

url = 'http://localhost:8000/predict'
try:
    r = requests.post(url, json={'X': X}, timeout=10)
    print('Status:', r.status_code)
    print('Response:', r.json())
except Exception as e:
    print('Request failed:', e)
```
