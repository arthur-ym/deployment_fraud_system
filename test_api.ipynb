# Test API Notebook
Calls the local FastAPI `/predict` endpoint using the first element from `artifacts/payload_test.json`.

```python
from pathlib import Path
import json

p = Path('../artifacts/payload_test.json') if Path('artifacts/payload_test.json').exists() else Path('artifacts/payload_test.json')
data = json.loads(p.read_text())

if isinstance(data, list):
    first = data[0]
elif isinstance(data, dict):
    # prefer key 'X' if present
    if 'X' in data:
        first = data['X']
    else:
        # take first value from the dict
        first = next(iter(data.values()))
else:
    first = data

# Normalize to a list suitable for the /predict endpoint
if isinstance(first, list):
    X = first
elif isinstance(first, (int, float)):
    X = [first]
elif isinstance(first, dict) and 'X' in first:
    X = first['X']
else:
    X = [first]

print('Using X =', X)
```

```python
import requests

url = 'http://localhost:8000/predict'
try:
    r = requests.post(url, json={'X': X}, timeout=10)
    print('Status:', r.status_code)
    print('Response:', r.json())
except Exception as e:
    print('Request failed:', e)
```
